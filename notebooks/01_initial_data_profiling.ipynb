## 01. Initial Data Profiling

--- 

### Objective: 
    
- The objective of this notebook is to perform a thorough initial exploration and profiling of all raw datasets. 

- This step helps us understand the structure, quality, and basic characteristics of the data before    moving to data cleaning, feature engineering, and analysis.

### Why Data Profiling is Needed?

Before cleaning or analyzing any data, it is essential to understand what the data looks like. Data Profiling helps us answer important questions such as:

- How many rows and columns does each table have?
- What are the data types of each column?
- Are there any missing values? If yes, how many and in which columns?
- Are there any duplicate records?
- Which columns can be used as Primary Keys?
- What is the date range of the transactions?

#### 1. Importing libraries


```python
## 1. Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from IPython.display import display 
```

#### 2. Load all Datasets & Set Display

```python
## 2. Set  Display options

pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',100)
pd.set_option('display.float_format','{:.2f}'.format)

## 3. Define path

raw_path="C:/Users/Rohit/Project Related to Data Analyst/Indian-E-commerce-Sales-and-Customer-Analytics/data/raw_data/"

# Create a list files we use 

files = {
    "customers" : 'customers.csv',
    "customer_reviews":'customer_reviews.csv',
    "orders":'orders.csv',
    "order_items":'order_items.csv',
    "products":'products.csv',
    "payments":'payments.csv',
    "returns":'returns.csv',
    "shipments":'shipments.csv'
}

## 4. Load all datasets

datasets = {}

for name,filename in files.items():
    file_path = os.path.join(raw_path,filename) 
    
    try:
        datasets[name] = pd.read_csv(file_path)
        print(f"Loaded: {name:15} - {datasets[name].shape[0]:,} rows * {datasets[name].shape[1]} columns")
        print(f"{name} datasets loaded successfully !\n") 
        
    except Exception as e:
        print(f"Failed to Load {filename}: {e}")

```

#### 3. Profiling Functions 

- Instead of writing same profiling code to eight dataframe sperately then it's Time consuming that why we process them using loop

```python
# Basic Profiling Function

def profiling_dataframe(name,df):
    print("="*80)
    print(f"Dataset:{name.upper()}")
    print("="*80)
    
    ## Basic Info of rows and columns 
    
    print(f"Shape: {df.shape[0]} rows * {df.shape[1]} columns\n")           ## Gives Total Rows and columns of each datasets.
    
    ## Display data types of Each Datasets
    
    print("Column Name and Data Types:")
    dtype_df = pd.DataFrame({
        "Column" : df.columns,
        "Data types" : df.dtypes.values,
        "Non-Null Count": df.count().values,
        "Null Count": df.isnull().sum().values,
        "Null %" : (df.isnull().sum().values / len(df) * 100).round(2)
    })
    display(dtype_df)
    
   ## Missing Values Summary
     
    print("Missing value:")
    missing = df.isnull().sum()                                             ## Count total Missing value of Each datasets.
    missing = missing[missing >0]
    if len(missing) > 0:
        print("Columns with Missing Values:")
        missing_df = pd.DataFrame({
        'Missing count': missing,
        'Missing %': (missing / len(df) * 100).round(2)
        }).sort_values("Missing %", ascending = False)
        display(missing_df)
    else:
        print("No Missing Values Found!")
        
    
    ## Finding Duplicate Rows of Each Datasets

    print("Duplicate Rows:",df.duplicated().sum())                          ## Finding Duplicates records 
    print("\n")
    
    ## Sample Data
    
    print("\n First 5 Rows:")
    display(df.head())                                                      ## Give First 5 rows of Each datasets
    
    print("\n Last 5 Rows:")
    display(df.tail())                                                      ## Give Last 5 rows of Each Datasets.

## Numerical Summary

    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if num_cols:
        print("\n📊 Numerical Columns Summary:")
        display(df[num_cols].describe().T)
        
## Categorical Columns - unique counts

    cat_cols = df.select_dtypes(include = ['str','category']).columns.tolist()
    if cat_cols:
        print("\n Categorical Columns - Unique Value Counts:")
        unique_counts = pd.DataFrame({
            "Column": cat_cols,
            "Unique Values": [df[col].nunique() for col in cat_cols],
            "Most Frequent": [df[col].mode()[0] if not df[col].mode().empty else None for col in cat_cols]
        })
        display(unique_counts.sort_values("Unique Values",ascending= False))
        
## Run Profiling on All Datasets

for name, df in datasets.items():
    profiling_dataframe(name, df)

```

 #### 4. Aditional Important Checks 

```python
## Aditional Imported Checks like primary key, Date range, value count

print("="*80)
print("Additional Important Checks")
print("="*80)

# ----------- Check Primary Key candidates ---------

print("\n Checking Potential Primary Keys (Uniqueness):\n")

key_checks = {
    "customers" : "customer_id",
    "products" : "product_id",
    "orders" : "order_id",
    "order_items": "order_item_id",
    "payments": "payment_id",
    "shipments": "shipment_id",
    "returns": "return_id",
    "customer_reviews": "review_id"
}

for table , key_col in key_checks.items():
    if table in datasets and key_col in datasets[table].columns:
        total = len(datasets[table])
        unique = datasets[table][key_col].nunique()
        is_unique = total == unique
        status = "Unique (Primary key)" if is_unique else " Not Unique"
        print(f"{table:15} - {key_col:20} | Total : {total:,} | Unique : {unique:,} | {status}")
    else:
        print(f"{table:15} - Column '{key_col}' not found (please check actual column name)")
        

# ----------- Date Range Check (Orders) ---------

print("\n Data Range Analysis:")

if "orders" in datasets:
    orders = datasets['orders']
    
    # Trying Comman date column
    date_cols = [col for col in orders.columns if 'date' in col.lower() or 'time' in col.lower()]
    print(f"possible date columns in orders: {date_cols}")
    
    for col in date_cols:
        try:
            orders[col] = pd.to_datetime(orders[col], dayfirst=True, errors='coerce')
            print(f"\n{col}")
            print(f"Min Date : {orders[col].min()}")
            print(f"Max Date : {orders[col].max()}")
            print(f"Null     : {orders[col].isnull().sum()}")
        except:
            print(f"Could not convert {col} to datetime")
            
```
